```python
import pandas as pd
import folium

from folium.plugins import HeatMap

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/processed/cleaned_accident_data.csv")

df.head()
```

```python
print(df.shape)

df[["city","latitude","longitude"]].head()
```

# Geospatial Analysis

## Objective

The objective of this analysis is to visualize the geographical distribution of road accidents using latitude and longitude coordinates. This helps identify accident-prone locations and supports data-driven road safety planning.

## Why is this important?

Traditional charts explain accident patterns, but geospatial visualization reveals *where* accidents are concentrated. These insights help authorities prioritize infrastructure improvements, traffic monitoring, and emergency response planning.

```python
india_map = folium.Map(
    location=[22.97,78.65],
    zoom_start=5,
    tiles="CartoDB positron"
)

india_map
```

```python
for _, row in df.iterrows():

    folium.CircleMarker(

        location=[row["latitude"], row["longitude"]],

        radius=2,

        color="red",

        fill=True,

        fill_color="red",

        fill_opacity=0.5

    ).add_to(india_map)

india_map
```

#  Accident HeatMap

```python
heat_data = df[["latitude", "longitude"]].values.tolist()

heat_data[:5]
```

```python
heat_map = folium.Map(
    location=[22.97, 78.65],
    zoom_start=5,
    tiles="CartoDB positron"
)

HeatMap(
    heat_data,
    radius=10,
    blur=8,
    max_zoom=10
).add_to(heat_map)

heat_map
```

# Accident Hotspot Heatmap

## Objective

The objective of this analysis is to identify accident hotspots by visualizing the geographical concentration of accidents across different locations.

## Observation

- Areas with darker heat intensity indicate a higher concentration of accidents.
- Accident hotspots are concentrated in specific urban regions rather than being uniformly distributed.
- Several cities exhibit significantly higher accident density.

## Business Insight

Hotspot analysis enables authorities to identify high-risk locations and prioritize road safety improvements, surveillance, and emergency response planning.

## Recommendation

- Increase traffic monitoring in hotspot regions.
- Improve road infrastructure at accident-prone locations.
- Install warning signs and intelligent traffic management systems in high-risk areas.
- Prioritize hotspot locations for future road safety initiatives.

```python
severity_colors = {
    "minor": "green",
    "major": "orange",
    "fatal": "red"
}

severity_map = folium.Map(
    location=[22.97, 78.65],
    zoom_start=5,
    tiles="CartoDB positron"
)

for _, row in df.iterrows():

    folium.CircleMarker(
        location=[row["latitude"], row["longitude"]],
        radius=3,
        color=severity_colors[row["accident_severity"]],
        fill=True,
        fill_color=severity_colors[row["accident_severity"]],
        fill_opacity=0.7,

        popup=f"""
        <b>City:</b> {row['city']}<br>
        <b>State:</b> {row['state']}<br>
        <b>Severity:</b> {row['accident_severity'].title()}<br>
        <b>Weather:</b> {row['weather']}<br>
        <b>Cause:</b> {row['cause']}
        """

    ).add_to(severity_map)

severity_map
```

```python
legend_html = """
<div style="
position: fixed;
bottom: 50px;
left: 50px;
width: 180px;
background-color: white;
border:2px solid grey;
z-index:9999;
padding:10px;
font-size:14px;
">

<b>Accident Severity</b><br><br>

🟢 Minor<br>
🟠 Major<br>
🔴 Fatal

</div>
"""

severity_map.get_root().html.add_child(folium.Element(legend_html))
severity_map
```
